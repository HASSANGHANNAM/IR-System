```python
# ============================================================
# المعالجة المسبقة (Preprocessing) - Stemming vs Lemmatization
# ============================================================

# 🔧 اضبط هذا السطر قبل التشغيل:
# - False  → يستخدم Stemming (أسرع)
# - True   → يستخدم Lemmatization (أدق، لكن أبطأ قليلاً)
USE_LEMMATIZATION = False  # ← غيّره إلى True للتشغيل الثاني

# ------------------------------------------------------------
# 1. تثبيت المكتبات (مرة واحدة)
# ------------------------------------------------------------
!pip install ir_datasets nltk tqdm -q

# ------------------------------------------------------------
# 2. الاستيرادات
# ------------------------------------------------------------
import ir_datasets
import nltk
import json
import string
import os
import unicodedata
from tqdm import tqdm
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag

# تحميل بيانات NLTK (مرة واحدة)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)   # ← هذا السطر الجديد
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)

# ------------------------------------------------------------
# 3. إعداد الأدوات
# ------------------------------------------------------------
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(treebank_tag):
    """تحويل POS tag من NLTK إلى صيغة WordNet"""
    tag_map = {
        'J': 'a',  # Adjective
        'N': 'n',  # Noun
        'V': 'v',  # Verb
        'R': 'r'   # Adverb
    }
    return tag_map.get(treebank_tag[0].upper(), 'n')

def clean_text(text):
    """
    تنظيف النص بالكامل:
    1. Normalization (Unicode)
    2. Lowercasing
    3. إزالة علامات الترقيم
    4. Tokenization
    5. إزالة Stopwords
    6. Stemming أو Lemmatization (حسب الاختيار)
    """
    if not isinstance(text, str):
        return []

    # 1. Normalization (توحيد الترميز)
    text = unicodedata.normalize('NFKC', text)

    # 2. Lowercasing
    text = text.lower()

    # 3. إزالة علامات الترقيم
    text = text.translate(str.maketrans('', '', string.punctuation))

    # 4. Tokenization
    tokens = word_tokenize(text)

    # 5. إزالة Stopwords
    tokens = [t for t in tokens if t not in stop_words]

    # 6. التطبيع النهائي (Stemming أو Lemmatization)
    if USE_LEMMATIZATION:
        # Lemmatization (مع POS tagging لتحسين الدقة)
        pos_tags = pos_tag(tokens)
        tokens = [lemmatizer.lemmatize(word, get_wordnet_pos(tag)) for word, tag in pos_tags]
    else:
        # Stemming (أسرع)
        tokens = [stemmer.stem(t) for t in tokens]

    return tokens

# ------------------------------------------------------------
# 4. تحميل مجموعة البيانات
# ------------------------------------------------------------
print("🔄 جاري تحميل البيانات...")
dataset = ir_datasets.load("beir/webis-touche2020")
print(f"✅ عدد الوثائق: {len(dataset.docs)}")
print(f"✅ عدد الاستعلامات: {len(dataset.queries)}")

# ------------------------------------------------------------
# 5. تحديد مجلد الإخراج (حسب الطريقة المختارة)
# ------------------------------------------------------------
method_name = "lemmatization" if USE_LEMMATIZATION else "stemming"
OUTPUT_DIR = f"/content/processed_{method_name}"
os.makedirs(OUTPUT_DIR, exist_ok=True)
print(f"📁 سيتم الحفظ في: {OUTPUT_DIR}")

# مسارات الملفات
ORIGINAL_FILE = os.path.join(OUTPUT_DIR, 'corpus_original.jsonl')
CLEANED_TITLE_FILE = os.path.join(OUTPUT_DIR, 'corpus_cleaned_title.jsonl')
CLEANED_TEXT_FILE = os.path.join(OUTPUT_DIR, 'corpus_cleaned_text.jsonl')
CLEANED_QUERIES_FILE = os.path.join(OUTPUT_DIR, 'queries_cleaned.jsonl')

# ------------------------------------------------------------
# 6. معالجة الوثائق (382,545 وثيقة)
# ------------------------------------------------------------
print(f"🔄 جاري معالجة الوثائق باستخدام {method_name}...")

with open(ORIGINAL_FILE, 'w', encoding='utf-8') as f_orig, \
     open(CLEANED_TITLE_FILE, 'w', encoding='utf-8') as f_title, \
     open(CLEANED_TEXT_FILE, 'w', encoding='utf-8') as f_text:

    for doc in tqdm(dataset.docs_iter(), total=len(dataset.docs), desc="معالجة الوثائق"):
        # استخراج الحقول
        doc_id = doc.doc_id
        title = doc.title or ""
        text = doc.text or ""

        # 1. حفظ النص الأصلي
        original_record = {
            'doc_id': doc_id,
            'title': title,
            'text': text
        }
        f_orig.write(json.dumps(original_record) + '\n')

        # 2. تنظيف العنوان
        cleaned_title = clean_text(title)
        title_record = {
            'doc_id': doc_id,
            'cleaned_tokens': cleaned_title
        }
        f_title.write(json.dumps(title_record) + '\n')

        # 3. تنظيف النص
        cleaned_text = clean_text(text)
        text_record = {
            'doc_id': doc_id,
            'cleaned_tokens': cleaned_text
        }
        f_text.write(json.dumps(text_record) + '\n')

print("✅ تمت معالجة الوثائق بنجاح!")

# ------------------------------------------------------------
# 7. معالجة الاستعلامات (49 استعلام)
# ------------------------------------------------------------
print(f"🔄 جاري معالجة الاستعلامات باستخدام {method_name}...")

with open(CLEANED_QUERIES_FILE, 'w', encoding='utf-8') as f_queries:
    for query in tqdm(dataset.queries_iter(), total=len(dataset.queries), desc="معالجة الاستعلامات"):
        qid = query.query_id
        query_text = query.text or ""

        cleaned_query = clean_text(query_text)

        query_record = {
            'query_id': qid,
            'original_text': query_text,
            'cleaned_tokens': cleaned_query
        }
        f_queries.write(json.dumps(query_record) + '\n')

print("✅ تمت معالجة الاستعلامات بنجاح!")
print(f"🎉 انتهت المعالجة باستخدام {method_name}! الملفات في: {OUTPUT_DIR}")
```

```python

!pip install ir_datasets nltk tqdm -q
import ir_datasets
import nltk
import json
import string
import os
import unicodedata
from tqdm import tqdm
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)
USE_LEMMATIZATION = True
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
def get_wordnet_pos(treebank_tag):
    tag_map = {
        'J': 'a',
        'N': 'n',
        'V': 'v',
        'R': 'r'
    }
    return tag_map.get(treebank_tag[0].upper(), 'n')
def clean_text(text):
    if not isinstance(text, str):
        return []
    text = unicodedata.normalize('NFKC', text)
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    tokens = word_tokenize(text)
    tokens = [t for t in tokens if t not in stop_words]
    if USE_LEMMATIZATION:
        pos_tags = pos_tag(tokens)
        tokens = [lemmatizer.lemmatize(word, get_wordnet_pos(tag)) for word, tag in pos_tags]
    else:
        tokens = [stemmer.stem(t) for t in tokens]
    return tokens
print("🔄 جاري تحميل البيانات...")
dataset = ir_datasets.load("beir/webis-touche2020")
print(f"✅ عدد الوثائق: {len(dataset.docs)}")
print(f"✅ عدد الاستعلامات: {len(dataset.queries)}")
method_name = "lemmatization" if USE_LEMMATIZATION else "stemming"
OUTPUT_DIR = f"/content/processed_{method_name}"
os.makedirs(OUTPUT_DIR, exist_ok=True)
print(f"📁 سيتم الحفظ في: {OUTPUT_DIR}")
ORIGINAL_FILE = os.path.join(OUTPUT_DIR, 'corpus_original.jsonl')
CLEANED_TITLE_FILE = os.path.join(OUTPUT_DIR, 'corpus_cleaned_title.jsonl')
CLEANED_TEXT_FILE = os.path.join(OUTPUT_DIR, 'corpus_cleaned_text.jsonl')
CLEANED_QUERIES_FILE = os.path.join(OUTPUT_DIR, 'queries_cleaned.jsonl')
print(f"🔄 جاري معالجة الوثائق باستخدام {method_name}...")
with open(ORIGINAL_FILE, 'w', encoding='utf-8') as f_orig, \
     open(CLEANED_TITLE_FILE, 'w', encoding='utf-8') as f_title, \
     open(CLEANED_TEXT_FILE, 'w', encoding='utf-8') as f_text:
    for doc in tqdm(dataset.docs_iter(), total=len(dataset.docs), desc="معالجة الوثائق"):
        doc_id = doc.doc_id
        title = doc.title or ""
        text = doc.text or ""
        original_record = {
            'doc_id': doc_id,
            'title': title,
            'text': text
        }
        f_orig.write(json.dumps(original_record) + '\n')
        cleaned_title = clean_text(title)
        title_record = {
            'doc_id': doc_id,
            'cleaned_tokens': cleaned_title
        }
        f_title.write(json.dumps(title_record) + '\n')
        cleaned_text = clean_text(text)
        text_record = {
            'doc_id': doc_id,
            'cleaned_tokens': cleaned_text
        }
        f_text.write(json.dumps(text_record) + '\n')
print("✅ تمت معالجة الوثائق بنجاح!")
print(f"🔄 جاري معالجة الاستعلامات باستخدام {method_name}...")
with open(CLEANED_QUERIES_FILE, 'w', encoding='utf-8') as f_queries:
    for query in tqdm(dataset.queries_iter(), total=len(dataset.queries), desc="معالجة الاستعلامات"):
        qid = query.query_id
        query_text = query.text or ""
        cleaned_query = clean_text(query_text)
        query_record = {
            'query_id': qid,
            'original_text': query_text,
            'cleaned_tokens': cleaned_query
        }
        f_queries.write(json.dumps(query_record) + '\n')
print("✅ تمت معالجة الاستعلامات بنجاح!")
print(f"🎉 انتهت المعالجة باستخدام {method_name}! الملفات في: {OUTPUT_DIR}")

```
