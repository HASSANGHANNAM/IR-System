```python
import json

titles = []
texts = []

with open('/content/processed_stemming/corpus_cleaned_title.jsonl', 'r') as f:
    for line in f:
        data = json.loads(line)
        titles.append(' '.join(data['cleaned_tokens']))

with open('/content/processed_stemming/corpus_cleaned_text.jsonl', 'r') as f:
    for line in f:
        data = json.loads(line)
        texts.append(' '.join(data['cleaned_tokens']))

print(f"✅ تم تحميل {len(titles)} عنوان و {len(texts)} نص.")
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
print("✅ تم تحميل النموذج")
print(f"   حجم المتجه: {model.get_sentence_embedding_dimension()}")
```

```python
import numpy as np
from tqdm import tqdm
import gc

def generate_embeddings(texts, model, batch_size=64, output_file='embeddings.npy'):
    embeddings = []
    total = len(texts)
    print(f"🔄 جاري توليد المتجهات لـ {total} نص...")

    for i in tqdm(range(0, total, batch_size)):
        batch = texts[i:i+batch_size]
        batch_embeddings = model.encode(batch, convert_to_numpy=True, show_progress_bar=False)
        embeddings.append(batch_embeddings)

        # تنظيف الذاكرة كل 10 دفعات
        if i % (batch_size * 10) == 0:
            gc.collect()

    all_embeddings = np.vstack(embeddings)
    np.save(output_file, all_embeddings)
    print(f"✅ تم حفظ {len(all_embeddings)} متجه في {output_file}")
    return all_embeddings

# توليد متجهات العناوين
print("\n📌 بدء توليد متجهات العناوين...")
title_embeddings = generate_embeddings(titles, model, output_file='/content/embeddings_title.npy')

# توليد متجهات النصوص
print("\n📌 بدء توليد متجهات النصوص...")
text_embeddings = generate_embeddings(texts, model, output_file='/content/embeddings_text.npy')
```

```python
import json
import pickle
from rank_bm25 import BM25Okapi
from tqdm import tqdm
import os

print("🔄 جاري تحميل النصوص المنظفة...")

# قراءة النصوص المنظفة
tokenized_texts = []  # قائمة بقوائم الكلمات لكل وثيقة
doc_ids = []

with open('/content/processed_stemming/corpus_cleaned_text.jsonl', 'r', encoding='utf-8') as f:
    for line in tqdm(f, desc="قراءة الوثائق"):
        data = json.loads(line)
        doc_ids.append(data['doc_id'])
        tokenized_texts.append(data['cleaned_tokens'])  # قائمة الكلمات

print(f"✅ تم تحميل {len(tokenized_texts)} وثيقة.")

# تدريب نموذج BM25
print("🔄 جاري تدريب نموذج BM25 (قد يستغرق 2-3 دقائق)...")
bm25 = BM25Okapi(tokenized_texts)

print("✅ تم تدريب النموذج بنجاح!")

# حفظ النموذج
print("🔄 جاري حفظ النموذج...")
model_data = {
    'bm25': bm25,
    'doc_ids': doc_ids,
    'avgdl': bm25.avgdl,
    'corpus_size': len(doc_ids)
}

with open('bm25_model.pkl', 'wb') as f:
    pickle.dump(model_data, f)

print("✅ تم حفظ النموذج في bm25_model.pkl")
print(f"   حجم الملف: {os.path.getsize('bm25_model.pkl') / (1024*1024):.2f} ميجابايت")
```
