```python
import ir_datasets

dataset = ir_datasets.load("beir/webis-touche2020")

print(f"✅ عدد الوثائق: {len(dataset.docs)}")
print(f"✅ عدد الاستعلامات: {len(dataset.queries)}")
print(f"✅ عدد التقييمات (qrels): {len(dataset.qrels)}")
```

```python
for doc in dataset.docs_iter():
    print(doc)
    break
```

```python
for query in dataset.queries_iter():
    print(query)
    break
```

```python
for qrel in dataset.qrels_iter():
    print(qrel)
    break
```
